# PaddleOcrVL inference

```python
from paddleocr import PaddleOCRVL
import time
```

```python
ocr = PaddleOCRVL()
```

Table

```python
now = time.time()

img_path = "/content/cropped_objects/images/table_0.png"
output = ocr.predict(input=img_path)

print("time: ", time.time() - now)
```

```python
parsing_res = output[0]["parsing_res_list"]
print(parsing_res)
```

```python
for value in parsing_res:
    print("label: ", value.label)
    print("content: ", value.content)
    print("type: ", type(value.content))
    print("\n")
```
